```python
import pandas as pd
import numpy as np

df = pd.read_csv("datasets/ETT-data/ETTh1_labeled.csv")

zeros = df.iloc[:, -1].eq(0).sum()
ones = df.iloc[:, -1].eq(1).sum()
total = len(df)
print(total)
print(ones)
red = (ones / total) * 100
num_ones = df["label"].sum()
print("% of extreme events:", red)
print("no. of extreme events:", num_ones)
```

```python
patch_size = 24

for i in range(0, len(df), patch_size):
    patch = df.iloc[i:i + patch_size]
    num_ones = patch.iloc[:, -1].eq(1).sum()
    if num_ones > 0:
        print(f"Patch {i//patch_size + 1} (rows {i}-{i+patch_size-1}): {num_ones} ones")
```

```python
patch_size = 24

ones_per_patch = (
    df.iloc[:, -1]
    .groupby(df.index // patch_size)
    .apply(lambda x: x.eq(1).sum())
)

# Filter out patches with 0 ones
ones_per_patch = ones_per_patch[ones_per_patch > 0]

# Group patch numbers by how many ones they contain
grouped = ones_per_patch.groupby(ones_per_patch).apply(lambda x: list(x.index))

for num_ones, patches in grouped.items():
    print(f"{len(patches)} Patches: {num_ones} one(s):")
```
